# SET
- mutable
- if we want to represent a group of unique value as a single entity then we should go for set
- duplicate not allowed
- insertion order not preserved but we can sort the element
- indexing and slicing not allowed for set
- heterogeneous elements are allowed
- we can represent set elements within curly brackets with comma sepretor
- we can apply methamaticaly operation like union ,intersection ,diffrence etc on set

```python
# union()
# intersection()
# difference()
# symmetric_difference()

x={10,20,30,40}
y={50,40}
print(x.union(y))
print(x | y)
print(x.intersection(y))
print(x & y)
print(x.difference(y))
print(x - y)
print(x.symmetric_difference(y))
print(x^y)
```
